```python
# Mobile Package Billing 18
def calculate_mobile(package,extra_gb):
  if package == 'basic':
    gb_cost= 1000
    free = 0
  elif package == 'standard':
    gb_cost = 1500
    free = 2

  elif package == 'premium':
    gb_cost= 2500
    free = 5

  else:
    print("Enter proper package")
  charged_data= extra_gb - free
  extra = charged_data * 100




  total = gb_cost + extra
  print("Package Price:",gb_cost)
  print("Free Data:",free,'GB')
  print("Charged Data:",charged_data,"GB")
  print("Extra Charges:",extra)
  print("Final Bill:",total)
calculate_mobile('standard',5)
```
